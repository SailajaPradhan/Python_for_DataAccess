```python
import requests
import pandas as pd
import sqlalchemy

# specify api url and headers in dictionary(key value) format
url= 'https://pokeapi.co/api/v2/pokemon/pikachu'
header = {
    "Content-Type": "application/json",
    "Accept-Encoding": "deflate"
}


# api request
response = requests.get(url, headers=header)
print(response.status_code)

# Convert response to JSON    
responseData = response.json()
# print(responseData)

# # normalize data and save to a df
df= pd.json_normalize(responseData)
df = pd.DataFrame([{
    'id': responseData['id'],
    'name': responseData['name'].capitalize(),
    'height': responseData['height'],
    'weight': responseData['weight']
}])
# print(df)

#create a sql server connection
engine = sqlalchemy.create_engine(
    'mssql+pyodbc://localhost\\SQLEXPRESS/Test_Sep26'
    '?driver=ODBC+Driver+17+for+SQL+Server'
    '&trusted_connection=yes'
)
# load df to sql table
df.to_sql(
    name='FactPokemon',
    con=engine,
    index=False,
    if_exists='replace',
)

print("Data loaded successfully!")
```
